```python
import polars as pl
df = pl.read_csv('dataset.csv', try_parse_dates=True)
df.head
```

```python
df.select(pl.all().null_count())
```

```python
df.describe()
```

```python
df = df.with_columns(
    (pl.col("SoC") * 100).round(0).cast(pl.Int64),
    pl.col("Battery_Current") + 80 
)
df
```

```python
df.describe()
```

```python
final_df = df.select(['Timestamp', 'SoC', 'Battery_Voltage', 'Battery_Current'])
final_df.head
```

```python
with open('dataset.csv', mode="w") as f:
    final_df.write_csv(f)
```
